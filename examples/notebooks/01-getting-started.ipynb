## 2. Basic PDF Ingestion

Let's ingest a sample PDF document. This creates text embeddings for semantic search.

```python
# Ingest a single PDF
!ragged ingest pdf ../sample_documents/data_visualization.pdf
```

**What happened:**
1. PDF quality was analysed (detects issues like scanned images, rotation)
2. Document was processed and chunked
3. Text embeddings were generated using Ollama
4. Chunks stored in ChromaDB for retrieval

## 5. Using Python API

For programmatic access, use ragged's Python API.

```python
# Import ragged components
from ragged.storage.vector_store import VectorStore
from ragged.retrieval.retriever import Retriever
from ragged.config.settings import get_settings

# Initialize settings
settings = get_settings()
print(f"ChromaDB URL: {settings.chroma_url}")
print(f"Ollama URL: {settings.ollama_url}")
```

```python
# Initialize vector store
store = VectorStore()

# Check document count
count = store.count()
print(f"Total chunks in database: {count}")
```

```python
# Perform semantic search
retriever = Retriever()
results = retriever.retrieve("data visualization best practices", k=3)

# Display results
for i, result in enumerate(results, 1):
    print(f"\nResult {i}:")
    print(f"Score: {result.get('score', 'N/A')}")
    print(f"Content: {result['document'][:200]}...")
    print(f"Metadata: {result.get('metadata', {})}")
```
